```python
import sklearn
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
```

```python
donnees = fetch_20newsgroups(subset="all", categories=["sci.med", "rec.autos"], remove = ("headers", "footers", "quotes"))
```

```python
print(len(donnees.data))
print(donnees.data[0])
print(donnees.target[0])
donnees.target_names[donnees.target[0]]
```

```python
import pandas as pd 
etiquettes = pd.Series(donnees.target)
print(etiquettes.value_counts())
etiquettes
```

```python
X_train, X_test, y_train, y_test = train_test_split(donnees.data, donnees.target, test_size=0.2, random_state=42, stratify=donnees.target)
print(len(X_train))
print(len(X_test))
print(len(y_train))
print(len(y_test))
y_test2 = pd.Series(y_test)
y_test2.value_counts()
```
