```python
import sklearn
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.dummy import DummyClassifier 
from sklearn.linear_model import LogisticRegression
```

```python
donnees = fetch_20newsgroups(subset="all", categories=["sci.med", "rec.autos"], remove = ("headers", "footers", "quotes"))
```

```python
print(len(donnees.data))
print(donnees.data[0])
print(donnees.target[0])
donnees.target_names[donnees.target[0]]
```

```python
import pandas as pd 
etiquettes = pd.Series(donnees.target)
print(etiquettes.value_counts())
etiquettes
```

```python
X_train, X_test, y_train, y_test = train_test_split(donnees.data, donnees.target, test_size=0.2, random_state=42, stratify=donnees.target)
print(len(X_train))
print(len(X_test))
print(len(y_train))
print(len(y_test))
y_test2 = pd.Series(y_test)
y_test2.value_counts()
```

```python
#Transformation des textes en nombres
```

```python
vectoriseur = CountVectorizer()
X_train_vec = vectoriseur.fit_transform(X_train)
X_test_vec = vectoriseur.transform(X_test)
```

```python
print(X_train_vec.shape)
print(X_test_vec.shape)
print(len(vectoriseur.get_feature_names_out()))
```

4ème partie : La baseline avec le DummyClassifer

Ici, nous utilisons un modèle un modèle naïf qui n’apprend que la classe la plus fréquente, sans lire les textes. Ainsi, lorsque nous disposerons du modèle entraîné, nous pourrons le comparer à ce modèle naïf (fondé sur la classe la plus fréquente) afin de valider la pertinence de nos résultats.

```python
#On crée notre modèle en lui demandant d'utiliser la strategy most frequent
#puis on l'entraine avec fit avec textes vectorisés du train et leurs etiquettes

baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train_vec, y_train)
```

```python
#Évaluation du modèle :
print(baseline.score(X_test_vec,y_test))
baseline.predict(X_test_vec[:10])
```

Le résultat est logique, le Dummy repond a chaque fois 0 (rec.autos) sans lire les textes, sachant que la X_test_vec contient la moitié de rec.autos on obtient logiquement 0.5

5ème partie : Je commence le vrai modèle (régression logistique)

```python
#Création et entrainement
modele = LogisticRegression(max_iter=1000)
modele.fit(X_train_vec, y_train)
```

```python
#Comparaison entre le score d'entrainement et celui de test 
print(modele.score(X_train_vec, y_train))
print(modele.score(X_test_vec, y_test))
```

```python
print(modele.predict(X_test_vec[:10]))
print(y_test[:10])
```

Nous pouvons voir ici en faisait .predict des 10 premiers vec de data la différence entre la prédiction de notre modele et le résultat qui est réellement dans y_test

Mais pourquoi notre modèle s'est-il trompé ? Regardons de plus prêt :

```python
print(X_test[2])
```

Nous pouvons voir que plusieurs mots peuvent être ambigus pour notre modèle (problem, test, heat, soft…). C'est peut-être ce qui explique son erreur.

```python
#Probabilités de chaque classe (voitures, médecine) pour les 5 premiers textes du test
print(modele.predict_proba(X_test_vec[:5]))
```

Semaine 2 : Métrique

Métrique

```python
#je range dans une variable les predictions de mon modele 
#pour pouvoir les comparer avec les réponses (y_test) et
#faire la matrice de confusion 
prediction = modele.predict(X_test_vec)
```

Nous voyons bien que la somme de la diagonale (VN + VP) donne 256, si on divise par le total du nombre d'échnatillon test, on retombe bien sur 0.89 

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import *

confusion_matrix(y_test, prediction)
```

Ici, précision_score et recall_score font eux meme le comtpage, VP et les FP (prediction) et des VP et FN (recall)

```python
print(precision_score(y_test, prediction))
print(recall_score(y_test, prediction))
```

f1_score combine la précision et le rappel dans un score a l'aide d'une formule 2*(P*R)/(P+R), il tire vers le plys gaible des deux donc si le reusltat est élevé c'est que la précision et le rappel sont bons

```python
f1_score(y_test, prediction)
```
